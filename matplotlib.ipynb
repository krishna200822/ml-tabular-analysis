```python
# Univariate-numerical
data ={
"Salary" :[20000, 25000, 55000, 30000, 89000, 35000, 38000, 40000, 80000, 45000, 48000, 50000, 55000, 60000, 5000]
        }
df = pd.DataFrame(data)
```

```python
#Univaitre categorical columns
```

```python
depts = ['hr', 'it', 'finance'] * (len(df) // 3)
df['dept'] = depts
```

```python
df.head()
```

```python
df.info()
```

```python
# Pie chart for department distribution
count = df['dept'].value_counts()
plt.pie(count, labels=count.index, autopct='%1.1f%%')
plt.title('Department Distribution')
plt.show()
```

```python
#count plot
plt.bar(count.index,count)
plt.show()
```

```python
# bivatre -num-num
```

```python
# Re-assigning ages to have a better correlation with salary
# Higher salaries will generally be paired with higher ages
# Salary sequence: [20k, 25k, 55k, 30k, 89k, 35k, 38k, 40k, 80k, 45k, 48k, 50k, 55k, 60k, 5k]
ages = [24, 26, 35, 28, 45, 30, 31, 32, 42, 33, 34, 36, 38, 40, 22]
df['Age'] = ages
display(df.head())
```

```python
# scatter plot - Bivariate analysis (Num-Num)
plt.scatter(df["Salary"], df["Age"], color="red", edgecolors='black')
plt.show()
```

```python
sort_age= df.sort_values("Age")

```

```python
#line plot
plt.plot(sort_age["Age"],df["Salary"], color="red",marker="o")
```

```python
# bar chart
plt.bar(sort_age["Age"],df["Salary"],color="green")
plt.show()
```

```python
# bivatire num-cat
```

```python
hr_sal = df[df["dept"] == 'hr']["Salary"]
hr_sal
```

```python
it_sal = df[df["dept"] == 'it']["Salary"]
it_sal
```

```python
finance_sal = df[df['dept'] == 'finance']['Salary']
finance_sal
```

```python
# box plot
plt.boxplot([hr_sal, it_sal, finance_sal], labels=['hr', 'it', 'finance'])
plt.legend()
plt.show()
```

```python
# pie chart
sal_by_dep= df.groupby("dept")["Salary"].sum()
sal_by_dep
```

```python
plt.pie(sal_by_dep, labels=sal_by_dep.index, autopct='%1.1f%%')
plt.axis("equal")
plt.show()
```

```python
# bar plot
hr_mean=sum(hr_sal)/len(hr_sal)
it_mean=sum(it_sal)/len(it_sal)
finance_mean=sum(finance_sal)/len(finance_sal)
```

```python
plt.bar(["hr",'it','finance'],[hr_mean,it_mean,finance_mean],color=["red","green","blue"])
plt.show()
```

```python
# Multivariate analysis: Salary, Age, and Experience
# Adding a sorted 'exp' (experience) column with 15 values between 1 and 10
df["exp"] = [1, 2, 2, 3, 4, 4, 5, 6, 7, 7, 8, 9, 9, 10, 10]
display(df.head())
```

```python
# bubble plot

plt.scatter(df["Salary"], df["Age"], s=df["exp"]*20, color="red")
```

```python
# 2 numerical and 1 catory
```

```python
# 2 numerical (Salary, Age) and 1 categorical (dept)
# Convert categorical 'dept' to codes for the color parameter 'c'
plt.scatter(df["Salary"], df["Age"], c=df["dept"].astype('category').cat.codes, s=100, cmap="rainbow", edgecolors='black')
plt.colorbar(label='Department Code')
plt.xlabel('Salary')
plt.ylabel('Age')
plt.title('Salary vs Age by Department')
plt.legend()
plt.show()
```
