```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
X, y_true = make_blobs(n_samples=500, centers=3, cluster_std=0.60, random_state=42
                       )
```

```python
X
```

```python
df =pd.DataFrame(X, columns=['feature 1', 'feature 2'])
```

```python
# The error `AttributeError: PathCollection.set() got an unexpected keyword argument 'paltte'` is in cell Qy5MXhIpOdE4.
# To fix it, change `paltte` to `palette` in the `sns.scatterplot` function call.
df
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
inertia =[]
k_range = range(1, 10)
for k in k_range:
    kmeans = KMeans(n_clusters= k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
inertia
```

```python
plt.plot(k_range, inertia, 'bx-')
plt.xlabel('k')
plt.ylabel('inertia')
plt.show()
#
```

```python
kmeans_final = KMeans(n_clusters=3, random_state=42)
```

```python
cluster_labels = kmeans_final.fit_predict(X_scaled)
```

```python
cluster_labels
```

```python
df ['cluster']= cluster_labels
```

```python
sns.scatterplot(x=df['feature 1'] ,
                y = df['feature 2'],
                hue = df['cluster'],
                palette = 'viridis')
plt.show()
#
```

```python
from sklearn.datasets import make_moons
```

```python
X , y = make_moons(n_samples=500, noise=0.05, random_state=42)
```

```python
from sklearn.cluster import KMeans ,DBSCAN
```

```python
df = pd.DataFrame(X, columns=['feature 1', 'feature 2'])
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans_labels = kmeans.fit_predict(X_scaled)
```

```python
df['kmeans_cluster'] = kmeans_labels
```

```python
sns.scatterplot(x=df['feature 1'] ,
                y = df['feature 2'],
                hue = df['kmeans_cluster'],
                palette = 'viridis')
plt.show()
#
```

```python
dbscan = DBSCAN(eps=0.2, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_scaled)
```

```python
df['dbscan_cluster'] = dbscan_labels
```

```python
sns.scatterplot(x=df['feature 1'] ,
                y = df['feature 2'],
                hue = df['dbscan_cluster'],
                palette = 'viridis')
plt.show()
#
```
